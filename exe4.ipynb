```python
L1=[22,445,36,50,44,"chaitu",2.5,'c']
L2=[1,5,"sravya","hasini"]
#slicing
print(L1[2:7])
#reversing
print(L1[::-1])
#concatenation
print(L1+L2)
#repeating
print(L2*3)
#membership operator
print("chaitu" in L1)
#iteration
for i in L2:
  print(i)
```
