```python
def palindrome(x):
  reverse=0
  original=x
  k=0
  while(x>0):
    k=x%10
    reverse=(reverse*10)+k
    x=x//10
  if(reverse==original):
    print("PALINDROME")
  else:
    print("NOT PALINDROME")
num=int(input())
palindrome(num)


```

```python
def reverse_str(x):
  rev=""
  for i in range (len(x)-1,-1,-1):
    rev=rev+x[i]
  return rev
print(reverse_str("hello"))
```

default argument
we give the value inside the function definition
keyword argument
we give the value in function calling
variable length argument
whenever we pass a collection of elements in the function definition it is stored in the tuple format

```python
#example for default arguments
def print_msg(name,branch="AI&DS"):
  print("name of the student is",name,"branch is",branch)
print_msg("chaitu")
```

```python
#example for keyword arguments
def print_msg(name,branch):
  print("name of the student is",name,"branch is",branch)
print_msg(branch="AI&DS",name="chaitu")
```

```python
#example for variable length arguments
def add(*num):
  print(num)
  print(sum(num))
add(1,2,3,4,5)
```

```python
#function that return multiple values
def result(*marks):
  total=sum((marks))
  n=len(marks)
  average=total/n
  return total,n,average
result_total,result_n,result_average=result(20,25,30,33,50)
print(f"total is {result_total} average is {result_average} total number of subjects is {result_n}")
```
